```python
import pandas as pd
import json
import sqlite3
import random
import logging

from faker import Faker
```

```python
fake = Faker()
```

```python
customers = []

for i in range(1000):

    customers.append({
        "CustomerID": i + 1,
        "Name": fake.name(),
        "Email": fake.email(),
        "Age": random.randint(18, 65),
        "City": fake.city()
    })

df = pd.DataFrame(customers)

df.head()
```

```python
df.shape
```

```python
df.to_csv("customers.csv", index=False)

print("Customer dataset saved successfully.")
```

```python
df.loc[5, "Email"] = None

df.loc[20, "Age"] = "Unknown"

df.loc[35, "CustomerID"] = None

df.to_csv("customers.csv", index=False)

df.head(40)
```

```python
invalid_age = df[df["Age"] == "Unknown"]

invalid_age
```

```python
bad_records = df[

    df["CustomerID"].isnull()

    |

    df["Email"].isnull()

    |

    (df["Age"] == "Unknown")

]

bad_records
```

```python
bad_records.to_csv("quarantine.csv", index=False)

print("Quarantine file created.")
```

```python
good_records = df.drop(bad_records.index)

good_records.head(40)
```

```python
transactions = []

for i in range(1500):

    transactions.append({

        "transaction_id": i + 1,

        "customer": {

            "id": random.randint(1, 200),

            "name": fake.name()

        },

        "payment": {

            "amount": random.randint(100, 5000),

            "mode": random.choice(["UPI", "Card", "Cash"])

        }

    })
```

```python
with open("api_logs.json", "w") as file:

    json.dump(transactions, file, indent=4)

print("JSON file created.")
```
